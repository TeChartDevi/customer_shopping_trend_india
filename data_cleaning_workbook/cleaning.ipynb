### 1. Null Value Check and Handling.

```python
# Null Value Check:
df.isnull().sum().sort_values(ascending=False)
```

Note:
- Found nulls in Column:
    - A. Festival/Sale             6788
    - B. Online Store              2249
    - C. Delivery Speed            2249
    - D. Size                      1035

##### A. Null Value Handling for Column Festival/Sale

```python
# 1. Column Unique Val check:
df['Festival/Sale'].unique()
```

```python
# 2. Null Val cross-relation check between 'Festival/Sale','Online Store':

df[df[['Festival/Sale','Online Store']].isnull().any(axis=1)][['Festival/Sale','Online Store']]
```

```python
# 3. Null Value Handling:

df['Festival/Sale'] = df['Festival/Sale'].fillna('Regular Day')
```

```python
# 4. Verifing Null Value Replacement:

print(df['Festival/Sale'].isna().sum())
```

##### B. Null Value Handling for Column Online Store:

```python
# 1. Null Value Handling:

df.loc[
    (df['Online/Offline'] == 'Offline') & 
    (df['Online Store'].isna()),
    'Online Store'
] = 'In-Store Purchase'

## Note: Where the Online/Offline was 'Offline' &  the Online Store was missing, I replaced the missing Online Store with "In-Store Purchase"
```

```python
# 2. Verifing Null Value Replacement:

df[['Online/Offline', 'Online Store']].head(10)
```

```python
# 3. Verifying any remaining Null:

df['Online Store'].isna().sum() 
```

##### C. Null Value Handling for Column Delivery Speed

```python
# C(1). Null Value Handling:
df.loc[
    (df['Delivery Time (Days)'] == 0) & 
    (df['Delivery Speed'].isna()), 
    'Delivery Speed'
] = 'N/A (Offline)'

## Note: Where the delivery time was 0 days &  the Delivery Speed was missing, I replaced the missing Delivery Speed with "N/A (Offline)"
```

```python
# C(2). Verifing Null Value Replacement:

print(df['Delivery Speed'].isna().sum())
```

##### D. Null Value Handling for Column Size

```python
# 1. Column Unique Val check:
df.loc[df['Size'].isna(), 'Category'].unique()
```

Note:
- All Null values are from the same category.

```python
# 2. Null Val cross-relation check between Category, Item Purchased & Size:

df[df[['Category', 'Item Purchased','Size']].isnull().any(axis=1)][['Category', 'Item Purchased', 'Size']]
```

Note:
- No specific sub-category relation found for the nulls. Thus, filling up with Unknown.

```python
# 3. Null value handling:

df['Size'] = df['Size'].fillna('Unknown')
```

```python
# 4.  Verifying Null Value Replacement:

print(df['Size'].isna().sum())
```

##### B. Renaming the Location column with Cities column.

##### C. Convert Purchase Date to relevant Dtype

```python
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'], errors = 'coerce')

print(df['Purchase Date'].dtype)
```

# Feature Engineering:
- A. Creating a New Region Column.
- B. Creating column for footwear and clothing size. 
- C. Extracting Parts of date.
- D. Creating Age group 

##### A. Creating a New Region Column:

```python
# 1. All unique city names:

df['Cities'].unique()
```

```python
# 2. Mapping the cities

region_map = {
    'Delhi': 'North',
    'Noida': 'North',
    'Gurgaon': 'North',
    'Chandigarh': 'North',
    'Jaipur': 'North',
    'Lucknow': 'North',

    'Chennai': 'South',
    'Hyderabad': 'South',
    'Visakhapatnam': 'South',
    'Kochi': 'South',
    'Bangalore': 'South',

    'Mumbai': 'West',
    'Pune': 'West',
    'Ahmedabad': 'West',
    'Nagpur': 'West',
    'Surat': 'West',

    'Kolkata': 'East',
    'Patna': 'East',
    'Bhubaneswar': 'East',

    'Indore': 'Central'
}

df['Region'] = df['Cities'].map(region_map)
```

```python
# 3. Popping out the region column:

region = df.pop('Region')

# Finding the position of the Cities column:

city_loc = df.columns.get_loc('Cities')

# Inserting Region column immediately after Cities column:

df.insert(city_loc + 1, 'Region', region)
```

##### B. Creating column for Footwear & Clothing Size. 

```python
# 1. Category column unique vals:
df['Category'].unique()
```

```python
# 2. Footwear Size:
Footwear_Size = np.where(df['Category'] == 'Footwear', df['Size'], 'Not Applicable')

# 3. Clothing Size:
Clothing_Size = np.where(df['Category'] != 'Footwear', df['Size'], 'Not Applicable')
```

```python
# 4. Finding Size position:

size_loc = df.columns.get_loc('Size')

# 5. Inserting columns immediately after Size column:

df.insert(size_loc + 1, 'Clothing_Size', Clothing_Size)

df.insert(size_loc + 2, 'Footwear_Size', Footwear_Size)
```

```python
# 7. Dropping Size column:
df.drop( columns = ['Size'], inplace = True)
```

```python
# 8. Column after changes:
df[['Category', 'Footwear_Size', 'Clothing_Size']].head(10)
```

##### C. Extracting Parts of Date

```python
# 1. Finding the position of the Purchase Date column:
purchase_date_loc = df.columns.get_loc('Purchase Date')


# 2. Extracting Year and inserting it immediately after the Purchase Date column:
df.insert(purchase_date_loc + 1, 'Year', df['Purchase Date'].dt.year)


# 3. Extracting Month and inserting it immediately after the Year column:
df.insert(purchase_date_loc + 2, 'Month', df['Purchase Date'].dt.month)


# 4. Extracting Day and inserting it immediately after the Month column:
df.insert(purchase_date_loc + 3, 'Day', df['Purchase Date'].dt.day)


# 5. Extracting Weekday Name and inserting it immediately after the Day column:
df.insert(purchase_date_loc + 4, 'Weekday', df['Purchase Date'].dt.day_name())
```

D. Creating Age group:

```python
def age_group(age):
    if age < 18:
        return 'Under 18'
    elif age <= 25:
        return '18-25'
    elif age <= 35:
        return '26-35'
    elif age <= 45:
        return '36-45'
    elif age <= 60:
        return '46-60'
    else:
        return 'Above 60'
    
# Finding the position of the age column:
age_loc = df.columns.get_loc('Age')

# Inserting the Age Group column immediately after the age column:
df.insert(age_loc + 1, 'Age Group', df['Age'].apply(age_group))
```
